# 02 — Driven star validation

This notebook regenerates the fully time-dependent anisotropic driven-star validation figures. It computes a small exact sparse/Krylov oracle and compares L0 and L0+G1 branch-continuous log errors.

```python
from pathlib import Path
import sys

# Works when the notebook is run from notebooks/ or from the repository root.
cwd = Path.cwd().resolve()
if cwd.name == "notebooks":
    ROOT = cwd.parent
else:
    candidates = [cwd] + list(cwd.parents)
    ROOT = next((p for p in candidates if (p / "experiments").is_dir() and (p / "src").is_dir()), cwd)

if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

FIGURES = ROOT / "figures"
FIGURES.mkdir(exist_ok=True)
print("Repository root:", ROOT)
print("Figures directory:", FIGURES)
```

```python
from experiments.run_driven_star_validation import driven_time_trace, driven_final_scaling

ts, err_l0, err_g1 = driven_time_trace()
ds, e0, e1, s0, s1 = driven_final_scaling()

print("Final time-trace errors:")
print("  L0   =", err_l0[-1])
print("  L0+G1=", err_g1[-1])
print("Degree scan:", ds)
print("L0 errors:", e0, "slope", s0)
print("L0+G1 errors:", e1, "slope", s1)
```

Generated outputs: `fig_driven_validation_error` and `fig_driven_validation_scaling` in PDF/PNG, with corresponding `.npz` data files.
