# 01 — Generate main paper figures

This notebook regenerates the core schematic and static-validation figures used in the manuscript. It calls `experiments/run_main_figures.py` through its public functions.

```python
from pathlib import Path
import sys

# Works when the notebook is run from notebooks/ or from the repository root.
cwd = Path.cwd().resolve()
if cwd.name == "notebooks":
    ROOT = cwd.parent
else:
    candidates = [cwd] + list(cwd.parents)
    ROOT = next((p for p in candidates if (p / "experiments").is_dir() and (p / "src").is_dir()), cwd)

if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

FIGURES = ROOT / "figures"
FIGURES.mkdir(exist_ok=True)
print("Repository root:", ROOT)
print("Figures directory:", FIGURES)
```

```python
from experiments.run_main_figures import (
    fig_star_primitive,
    fig_theory_schematic,
    fig_aligned_benchmark,
    fig_one_over_d_scaling,
    fig_static_inhomogeneous_validation,
)

fig_star_primitive()
fig_theory_schematic()
fig_aligned_benchmark()
fig_one_over_d_scaling()
fig_static_inhomogeneous_validation()
print("Generated main figures in", FIGURES)
```

Generated outputs: `fig_star_primitive`, `fig_theory_schematic`, `fig_aligned_benchmark`, `fig_one_over_d_scaling`, and `fig_static_inhom_validation` in both PDF and PNG formats where applicable. Numerical data are saved as `.npz` files for validation figures.

```python
for path in sorted(FIGURES.glob("fig_*")):
    print(path.relative_to(ROOT))
```
