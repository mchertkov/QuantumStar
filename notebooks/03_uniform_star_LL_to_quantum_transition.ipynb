# 03 — Uniform star LL-to-quantum transition

This notebook regenerates the homogeneous two-population leaf experiment used to illustrate the LL-classical validity horizon and the transition to the discrete quantum-sector regime.

```python
from pathlib import Path
import sys

# Works when the notebook is run from notebooks/ or from the repository root.
cwd = Path.cwd().resolve()
if cwd.name == "notebooks":
    ROOT = cwd.parent
else:
    candidates = [cwd] + list(cwd.parents)
    ROOT = next((p for p in candidates if (p / "experiments").is_dir() and (p / "src").is_dir()), cwd)

if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

FIGURES = ROOT / "figures"
FIGURES.mkdir(exist_ok=True)
print("Repository root:", ROOT)
print("Figures directory:", FIGURES)
```

```python
import matplotlib.pyplot as plt
from experiments.run_uniform_star_LL_to_quantum import (
    generate_uniform_star_transition_data,
    plot_uniform_star_transition,
)

data = generate_uniform_star_transition_data()
fig = plot_uniform_star_transition(data, save=True)
plt.show()

print("Representative degree:", int(data["rep_d"]))
print("L0 horizon:", float(data["rep_h_l0"]))
print("L0+G1 horizon:", float(data["rep_h_g1"]))
print("Degree scan:", data["ds"])
print("L0 horizons:", data["horizon_l0"])
print("L0+G1 horizons:", data["horizon_g1"])
```
